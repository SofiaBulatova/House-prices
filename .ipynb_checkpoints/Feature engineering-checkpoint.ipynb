```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
train = pd.read_csv('Data/train.csv')
test = pd.read_csv('Data/test.csv')
```

```python
missing = train.isnull().sum()
missing = missing[missing > 0].sort_values(ascending=False)
percent = (missing / len(train)) * 100
missing_data = pd.concat([missing, percent], axis=1, keys=['Total', 'Percent'])
missing_data.head(20)
```

Удаляем фичи с большим количеством пропусков (больше 2), удаляем строку с пустым значением Electrical

```python
columns = ['PoolQC', 'MiscFeature', 'Alley', 'Fence', 'MasVnrType', 'FireplaceQu', 'LotFrontage', 'GarageType', 'GarageYrBlt', 'GarageFinish', 
           'GarageQual', 'GarageCond', 'BsmtExposure', 'BsmtFinType2', 'BsmtQual', 'BsmtCond', 'BsmtFinType1', 'MasVnrArea' ]
train = train.drop(columns, axis=1)
train = train.dropna(subset=['Electrical'])
```

```python
train.isnull().sum().max()
```

```python
mis = test.isnull().sum()
mis = mis[mis > 0].sort_values(ascending=False)
perc = (mis / len(test)) * 100
missing_test = pd.concat([mis, perc], axis=1, keys=['Total', 'Percent'])
missing_test.head(20)
```

```python
test = test.drop(columns, axis=1)
```

```python
data = pd.concat([train['SalePrice'], train['GrLivArea']], axis=1)
data.plot.scatter(x='GrLivArea', y='SalePrice', ylim=(0,800000));
```

```python
train.sort_values(by = 'GrLivArea', ascending = False)[:2]
```

Удаляем два выброса GrLivArea по целевой переменной. С большой площадью дома и низкой стоимостью

```python
df_train = train.drop(train[train['Id'] == 1299].index)
df_train = train.drop(train[train['Id'] == 524].index)
```

```python
test.isnull().sum().sort_values(ascending=False)
```

Зполняем пустые значения в test на основе train

```python
train['MSZoning'].value_counts()
```

```python
test['MSZoning'] = test['MSZoning'].fillna('RL')
```

```python
train['Utilities'].value_counts()
```

```python
test['Utilities'].value_counts()
```

```python
df_train = df_train.drop(['Utilities'], axis=1)
test = test.drop(['Utilities'], axis=1)
```

```python
train['BsmtHalfBath'].value_counts()
```

```python
test['BsmtHalfBath'] = test['BsmtHalfBath'].fillna(0)
```

```python
test.isnull().sum().sort_values(ascending=False)
```

```python
train['BsmtFullBath'].value_counts()
```

```python
test['BsmtFullBath'] = test['BsmtFullBath'].fillna(0)
```

```python
train['Functional'].value_counts()
```

```python
test['Functional'] = test['Functional'].fillna('Typ')
```

```python
test.isnull().sum().sort_values(ascending=False)
```

```python
cols_zero_fill = ['BsmtFinSF1', 'BsmtFinSF2', 'BsmtUnfSF', 'TotalBsmtSF', 'GarageArea', 'GarageCars']
for col in cols_zero_fill:
    test[col] = test[col].fillna(0)
cat_cols_fill = ['Exterior1st', 'Exterior2nd', 'SaleType', 'KitchenQual']
for col in cat_cols_fill:
    mode_val = df_train[col].mode()[0]
    test[col] = test[col].fillna(mode_val)
```

```python
test.isnull().sum().sort_values(ascending=False)
```

Объединяем датасеты test train для one hot encoding и затем разъединяем

```python
train['_source'] = 'train'
test['_source'] = 'test'
combined = pd.concat([train, test], axis=0, sort=False)
object_cols = combined.select_dtypes(include=['str']).columns.tolist()
object_cols = [col for col in object_cols if col != '_source']
combined_encoded = pd.get_dummies(combined, columns=object_cols, drop_first=True)
train_encoded = combined_encoded[combined_encoded['_source'] == 'train'].drop(columns=['_source'])
test_encoded = combined_encoded[combined_encoded['_source'] == 'test'].drop(columns=['_source'])
```

```python
train_encoded = train_encoded.drop(['Id'], axis=1)
```

```python
train_encoded.head()
```

```python
train_encoded.to_csv('Data/train_processed.csv', index=False)
test_encoded.to_csv('Data/test_processed.csv', index=False)
```
